```python
#Check palindeome
s="madam"

if s==s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
    
```

```python
#Find factorial
n=5
fact=1

for i in range(1,n+1):
    fact*=i
print(fact)
```

```python
#Fibonacci series
n=10

a=0
b=1

for i in range(n):
    print(a,end="")
    a,b=b,a+b
```

```python
#Check Prime Number
n=3

if n<2:
    print("Not Prime")
else:
    for i in range(2,n):
        if n %i==0:
            print("Not Prime")
            break
        else:
            print("Prime")
       
```

```python
#Count Vowels
s="Python Programming"

vowels="aeiou"
count=0

for ch in s.lower():
    if ch in vowels:
        count+=1

        print(count)
```

```python
#Find Duplicate Values
numbers=[1,2,3,2,3,5]

duplicates=[]

for x in numbers:
    if numbers.count(x)>1 and x not in duplicates:
        duplicates.append(x)

print(duplicates)
```

```python
#find second largest number
numbers=[10,20,5,40,30]

numbers=list(set(numbers))
numbers.sort()

print(numbers[-2])
```
