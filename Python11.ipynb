```python
#Implicit Type casting
a=123
b=1.23
print(type(a))
print(type(b))

c=a+b
print(c)
print(type(c))
```

```python
#Explicit Type casting
a="123"
b=1.23
print(type(a))
print(type(b))

a=int(a)#Conversion
print("After Conversion",type(a))

c=a+b
print(c)
print(type(c))
```
