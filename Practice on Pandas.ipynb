Practice Questions in Pandas:-

```python
import pandas as pd

data={
    "Name":['Amit',"Sneha","Rahul","Pooja","Rohit"],
    "Age":[21,22,20,23,21],
    "Marks":[78,85,67,92,74],
    "City":["Pune","Mumbai","Pune","Nashik","Mumbai"]
}
df=pd.DataFrame(data)
print(df)
```

```python
#1.First 3 rows
print("First 3 rows:")
print(df.head(3))
```

```python
#2.Last 2 rows
print("Last 2 rows:")
print(df.tail(2))
```

```python
#3.Name and Marks columns
print("Name and Columns:")
print(df[["Name","Marks"]])
```

```python
#4.Marks>75 students
print("Students having marks greter than 75:")
print(df[df["Marks"]>75])
```

```python
#5.Average marks
print("Average Marks:")
print(df["Marks"].mean())
```
